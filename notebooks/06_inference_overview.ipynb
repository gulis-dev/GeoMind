# What is inference?

**Inference** is the stage where a trained artificial intelligence model is used to make predictions on new, previously unseen data.


## How does it work in the GeoMind project?

1. **Training:** First, the model is trained on a large dataset where the correct answers are known (e.g., images labeled with their regions and geographic coordinates).
2. **Inference:** Once the model is trained, you use it to make predictions on new images—that is, to determine which region and location the model predicts for an image it has never seen before.

## Why do we do inference?

- To use the model in real-world tasks—for example, to automatically assign a location to images, detect objects, recognize text, etc.
- Inference is fast—the model no longer learns, but only uses its fixed knowledge to make predictions.

## Summary

**Inference** simply means “using” a trained model on new data to obtain predictions.
In your project, this means automatically determining the geographical location based on an image.
